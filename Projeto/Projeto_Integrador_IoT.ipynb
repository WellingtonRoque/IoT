# 🌐 Introdução – O que é IoT?

## 📡 Internet das Coisas (IoT)

IoT (Internet of Things), ou Internet das Coisas, é o conceito de conectar dispositivos físicos à internet, permitindo que sensores, atuadores e sistemas troquem informações em tempo real.

A IoT possibilita:

- Monitoramento remoto
- Automação de processos
- Comunicação entre dispositivos
- Coleta de dados
- Processamento inteligente
- Tomada de decisão automática

Atualmente a IoT está presente em praticamente todas as áreas da tecnologia e da indústria.

---

# 🧠 Funcionamento de um Sistema IoT

Um sistema IoT normalmente é composto por:

```text
Sensores → Microcontrolador → Rede → Interface → Usuário
```

## 📌 Exemplo prático

1. Um sensor mede temperatura
2. O ESP32 processa os dados
3. O ESP32 envia informações para uma interface web
4. O usuário acompanha os dados remotamente
5. O sistema pode tomar decisões automaticamente
