# 1. Sinais Analógicos vs Digitais

## 🔘 Sinal Digital
- Apenas dois estados: 0 ou 1  
- Exemplo: botão pressionado ou não  

## 🎚️ Sinal Analógico
- Valores contínuos  
- Exemplo: 0V a 5V  

## 🏭 Exemplos industriais

- Temperatura  
- Pressão  
- Vazão  
- Corrente  

👉 Problema: o Arduino entende apenas valores digitais.


# 2. Conversor Analógico-Digital (ADC)

O ADC converte uma tensão analógica em um valor numérico.

## 🔢 Características (Arduino Uno)

- Resolução: 10 bits  
- Total de níveis: 1024  
- Faixa: 0 a 1023  

## 📐 Fórmula de conversão

```
V = (ADC * 5) / 1023
```

## 📊 Interpretação prática

| ADC | Tensão |
|-----|--------|
| 0   | 0V     |
| 512 | ~2.5V  |
| 1023| 5V     |


````python
// Leitura simples do ADC
int sensor = A0;

void setup() {
  Serial.begin(9600);
}

void loop() {
  int valor = analogRead(sensor);
  Serial.print("ADC: ");
  Serial.println(valor);
  delay(1000);
}
````

````python
// Conversão para tensão
int sensor = A0;

void setup() {
  Serial.begin(9600);
}

void loop() {
  int adc = analogRead(sensor);
  float tensao = (adc * 5.0) / 1023.0;

  Serial.print("Tensão: ");
  Serial.println(tensao);

  delay(1000);
}
````

# 3. Sensor LM35

## 📌 Característica

```
10 mV = 1 °C
```

## 📐 Conversão

```
Temperatura = tensão / 0.01
```


````python
// Leitura de temperatura com LM35
int sensor = A0;

void setup() {
  Serial.begin(9600);
}

void loop() {
  int adc = analogRead(sensor);
  float tensao = (adc * 5.0) / 1023.0;
  float temperatura = tensao / 0.01;

  Serial.print("Temperatura: ");
  Serial.println(temperatura);

  delay(1000);
}
````
