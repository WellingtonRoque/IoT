# 🚀 3. Hello Server (Primeiro contato)

Objetivo: Criar servidor que retorna o IP



````
#include <WiFi.h>
#include <WebServer.h>

const char* ssid = "Wokwi-GUEST";
const char* password = "";

WebServer server(80);

void handleRoot(){
  server.send(200,"text/plain","IP: " + WiFi.localIP().toString());
}

void setup(){
  Serial.begin(115200);
  WiFi.begin(ssid,password);

  while(WiFi.status()!=WL_CONNECTED) delay(500);

  Serial.println(WiFi.localIP());

  server.on("/",handleRoot);
  server.begin();
}

void loop(){
  server.handleClient();
}

````

# 💡 4. Controle de LED (Máquina)

Agora vamos controlar uma máquina via web



````
int led = 2;

void on(){
  digitalWrite(led,HIGH);
  server.sendHeader("Location","/");
  server.send(303);
}

void off(){
  digitalWrite(led,LOW);
  server.sendHeader("Location","/");
  server.send(303);
}

````

# 🔥 11. SISTEMA COMPLETO FINAL



````
#include <WiFi.h>
#include <WebServer.h>

const char* ssid="Wokwi-GUEST";
const char* password="";

WebServer server(80);

int m1=2,m2=4,m3=5;
int sensor=34;

float gerarTemperatura(){ return random(150,400)/10.0; }

String pagina(){

 float temp=gerarTemperatura();
 int adc=analogRead(sensor);

 String html="<!DOCTYPE html><html><head>";
 html+="<meta name='viewport' content='width=device-width'>";
 html+="<link href='https://cdn.jsdelivr.net/npm/bootstrap@5.3.0/dist/css/bootstrap.min.css'>";
 html+="</head>";

 html+="<body class='bg-dark text-white text-center'>";
 html+="<div class='container mt-5'>";

 html+="<h1>🏭 Sistema IoT</h1>";
 html+="<h2>Temp: "+String(temp)+"</h2>";
 html+="<h3>ADC: "+String(adc)+"</h3>";

 html+="<a href='/m1on' class='btn btn-success m-2'>M1</a>";
 html+="<a href='/m1off' class='btn btn-danger m-2'>OFF</a><br>";

 html+="</div></body></html>";

 return html;
}

void root(){ server.send(200,"text/html",pagina()); }

void setup(){
 Serial.begin(115200);
 pinMode(m1,OUTPUT);

 WiFi.begin(ssid,password);
 while(WiFi.status()!=WL_CONNECTED) delay(500);

 Serial.println(WiFi.localIP());

 server.on("/",root);
 server.begin();
}

void loop(){ server.handleClient(); }

````
